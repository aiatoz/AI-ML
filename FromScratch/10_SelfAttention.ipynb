### 1. Self Attention - Math
- Objective to implement scaled dot product attention

```python
import numpy as np

rndGen = np.random.default_rng()
```

```python
Q = rndGen.random((5,5))
K = rndGen.random((5,5))
V = rndGen.random((5,5))
dK = K.shape[0] #In real life this wont be a sq matrix, we will have to take shape[1], So K's dim will be (number of words, dimK)
```

```python
def softmax(X):
    X = X - np.max(X, axis=1, keepdims=True)

    expX = np.exp(X)

    return expX / np.sum(expX, axis=1, keepdims=True)
```

```python
sftOut = softmax(np.dot(Q,K.T) / np.sqrt(dK))
SelfAttention = np.dot(sftOut, V)
```

```python
print(SelfAttention)
```

### 2. Self Attention - Real Life Scenario
- Objective to implement scaled dot product attention, close to the real life scenario

```python
import numpy as np

np.random.seed(42)
tokens = ["The", "cat", "sat", "on", "the", "mat"]

seqLen = len(tokens)
embdDim = 8

# Random embeddings
embeddings = np.random.randn(seqLen, embdDim)
embeddings.shape
```

```python
wQ = np.random.randn(embdDim, embdDim)
wK = np.random.randn(embdDim, embdDim)
wV = np.random.randn(embdDim, embdDim)

Q = embeddings @ wQ
K = embeddings @ wK
V = embeddings @ wV
```

```python
Q.shape
```

```python
dK = K.shape[1]

scores = Q @ K.T / np.sqrt(dK)

print(scores.shape)
```

```python
attention_weights = softmax(scores)

np.set_printoptions(precision=3, suppress=True)
print(attention_weights)
```
