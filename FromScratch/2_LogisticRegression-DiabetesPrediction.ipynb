## Diabetes Prediction using Logistic Regression

```python
from sklearn.datasets import load_diabetes
import numpy as np

diabetes = load_diabetes() #Part of sklearn library
X = diabetes.data
y = diabetes.target #Its not a classification dataset, instead a value 
```

```python
y.shape
```

```python
updatedY = np.zeros(442)
Y = (y - min(y))/(max(y)-min(y))
```

```python
updatedY = [1 if i > 0.5 else 0 for i in Y]
y = updatedY #Classification data
```

```python
print(X)
print(X.shape)
```

So 442 rows and 10 features

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test = train_test_split(X,y,train_size=0.7,random_state=42)
```

```python
import numpy as np

randomGen = np.random.default_rng()

W = randomGen.random((10,))
b = 0.0
```

```python
def sigmoid(z):
    activation = 1/(1+ np.exp(-z))
    return activation
```

```python
def forwardPass(X, W, b):
    z = X @ W + b
    y_cap = sigmoid(z)
    return y_cap
```

```python
def BCE(y,y_cap):
    return -np.mean(y*np.log(y_cap)+(np.ones(np.shape(y))-y)*np.log(np.ones(np.shape(y_cap))-y_cap))
```

```python
def gradCompute(X,y,y_cap):
    
    #print(X.shape)
    #print(y.shape)
    #print(y_cap.shape)
    
    N = len(X)
    
    dw = (1/N)*(X.T @ (y_cap-y)) #Its not 2/N
    db = (1/N)*np.sum(y_cap-y)

    return dw,db
```

```python
epochs = 10000
lr = 0.05
```

```python
for i in range(epochs):
    y_cap = forwardPass(X_train, W, b)
    
    Loss = BCE(y_train, y_cap)
    
    dw, db = gradCompute(X_train, y_train, y_cap)

    W = W - lr*dw
    b = b - lr*db

    if(i%100==0):
        print(f"Epoch : {i}, Loss : {Loss}")
```

```python
y_test_pred = forwardPass(X_test, W, b)  
```

```python
Loss = BCE(y_test, y_test_pred)
```

```python
Loss
```

```python
tstPred = (y_test_pred >= 0.5).astype(int)
```

```python
print(f"Test Loss: {Loss:.4f}")
print(f"Accuracy: {np.mean(tstPred == y_test):.4f}")
```

```python
for i in range(10):
    print(f"Pred = {tstPred[i]}, Prob = {y_test_pred[i]:.3f}, GT = {y_test[i]}")
```
